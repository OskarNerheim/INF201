# What is a complex number?

A complex number is a number of the form a + bi where:
- a is the real part
- b is the imaginary part
- i is the imaginary unit with i² = −1

Complex numbers can also be expressed in polar form as r·e^{iθ} (magnitude r and angle θ). Common operations include addition, subtraction, multiplication, division, magnitude (|z|), and conjugation (a − bi).

# Why use a custom Complex class in Python?

- Encapsulation: stores real and imaginary parts together with a clear API (methods like conjugate, magnitude, etc.).
- Readability: named methods and readable string/representation improve code clarity.
- Operator overloading: implement +, -, *, / so instances behave like numbers in expressions.
- Validation and invariants: enforce types/constraints and provide helpful error messages.
- Extensibility: add domain-specific utilities (polar conversion, phase normalization, formatting).
- Testability: easier to write unit tests for custom behavior than for ad-hoc tuples or pairs.

Note: Python has a built-in complex type. A custom class is useful when you need extra methods, specific formatting, or different semantics. If you keep the module named `complex.py`, import carefully (e.g. `import complex as cmpl`) to avoid shadowing the builtin.

### Imports and creating Complex objects
This cell imports the local `complex` module (aliased to `cmpl`) and creates two `Complex` instances: `a` and `b`.
- `importlib.reload(cmpl)` ensures the latest local edits are used when re-running the notebook.
- `a` and `b` are used by subsequent cells, so run this cell first.

```python
# Cell 1: import and create objects
# Import the module using an alias to avoid shadowing the builtin type
import importlib
import complex as cmpl
importlib.reload(cmpl)

# Create two Complex instances
a = cmpl.Complex(2, -3)
b = cmpl.Complex(2, 3)

print('Created a and b')
print('a ->', a)
print('b ->', b)
```

### String representation and `repr()`
This cell shows how the class formats complex numbers for humans (`__str__`) and for developers (`__repr__`).
Use `repr()` to obtain a string that can help recreate the object programmatically.

```python
# Cell 2: string and repr examples
print('str(a):', str(a))
print('repr(a):', repr(a))

# recreate from parts to show repr usefulness
c = cmpl.Complex(a.re, a.im)
print('c (from parts) ->', c)
print('c == a ->', c == a)
```

### Basic arithmetic (Complex with Complex)
This cell demonstrates addition, subtraction, multiplication and division between two `Complex` instances.
Results are `Complex` instances which you can inspect with `repr()` if needed.

```python
# Cell 3: basic arithmetic (Complex with Complex)
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)
```

### Mixing with Python numbers
Shows how the class handles operations between `Complex` instances and Python `int`/`float` types, both `Complex + number` and `number + Complex`.
Reverse operators (`__radd__`, `__rmul__`) make `2 + a` and `3 * a` work.

```python
# Cell 4: mixing with Python numbers (both orders)
print('a + 2 =', a + 2)
print('2 + a =', 2 + a)
print('a * 3 =', a * 3)
print('3 * a =', 3 * a)
print('a - 1.5 =', a - 1.5)
print('5 - a =', 5 - a)
```

### Subtraction examples
Illustrates `Complex - Complex`, `Complex - number`, and `number - Complex`. Reverse subtraction (`__rsub__`) ensures both orders work.

```python
# Cell 5: subtraction examples (kept id)
print('Complex - Complex ->', cmpl.Complex(2, -3) - cmpl.Complex(2, 3))
print('Complex - float ->', cmpl.Complex(0, 1) - 1)
print('float - Complex ->', 1 - cmpl.Complex(0, 1))
```

### Division and equality
Demonstrates `Complex / Complex` and the equality operators. Also shows a safe example for division-by-zero using try/except.

```python
# Cell 6: division and equality examples (kept id)
print('a / b ->', cmpl.Complex(2, -3) / cmpl.Complex(2, 3))
print('b / a ->', cmpl.Complex(2, 3) / cmpl.Complex(2, -3))
print('Equality checks ->', cmpl.Complex(2, -3) == cmpl.Complex(2, -3), cmpl.Complex(1, 2) != cmpl.Complex(1, 3))
# Comparing with non-Complex returns False per implementation
print('Compare with int ->', cmpl.Complex(2, -3) == 2)

# Cell 7: division-by-zero demonstration
try:
    print('a / Complex(0,0) ->', a / cmpl.Complex(0, 0))
except ZeroDivisionError as e:
    print('Division by zero raised:', type(e).__name__, e)
```
